```python
import numpy as np
from typing import List, Tuple, Dict, Callable, Union
```

```python
def split(t: np.ndarray)-> Tuple[np.ndarray, int, np.ndarray]:
    pivot = t[0]
    a=[]
    b=[]
    for i in range(1,len(t)):
        if t[0]>t[i]:
            a.append(t[i])
        else:
            b.append(t[i])
    return (a,pivot,b)
```

```python
t=[4,6,2,3,7,1,8]
print(split(t))
```

```python
def qsel(t: np.ndarray, k: int)-> Union[int, None]:
    try:
        if len(t)==1 and k==0:
            return t[0]
        a,mid,b=split(t)
        
        if (k==len(a)):
            return mid
        if (k<len(a)):
            return qsel(a,k)
        elif(k>len(a)):
            return qsel(b,k-len(a)-1)
    except:
        return 
```

```python
t=[4,6,2,3,7,1,8]
print(qsel(t,4))
```

```python
def qsel_nr(t: np.ndarray, k: int)-> Union[int, None]:
    aux=t.copy()
    while len(aux)>0:
        if (len(aux)==1 and k==0):
            return aux[0]
        
        a,mid,b=split(aux)

        if (k==len(a)):
            return mid
        elif (k<len(a)):
            aux=a
        elif(k>len(a)):
            aux=b
            k-=(len(a)+1)
    return
```

```python
t=[4,6,2,3,7,1,8]
print(qsel_nr(t,4))
```

```python
t=list(range(11))[ : : -1]
t= np.random.permutation(11)
shift = 0
for k in range(len(t)):
    print(k, qsel(t, k+shift))
    print(k, qsel_nr(t, k+shift))
```
